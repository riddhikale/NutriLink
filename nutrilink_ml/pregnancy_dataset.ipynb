```python
import pandas as pd
import numpy as np
import random
```

```python
def generate_pregnancy_dataset(rows=500, seed=42):
    random.seed(seed)
    np.random.seed(seed)

    data = []

    for _ in range(rows):

        # Trimester (1–3)
        trimester = random.choice([1, 2, 3])

        # Weight range based on trimester
        base_weight = np.random.normal(50 + trimester * 2, 4)
        weight = round(base_weight, 2)

        # Hemoglobin realistic range (8–13)
        hemoglobin = round(np.random.normal(11.2, 1.2), 2)

        # Blood Pressure realistic
        systolicBP = random.randint(100, 140)
        diastolicBP = random.randint(60, 90)

        # Symptom Flags
        dizziness = random.choice([0, 1])
        fatigue = random.choice([0, 1])
        swelling = random.choice([0, 1])
        lowAppetite = random.choice([0, 1])

        # Past anemia history
        pastAnemia = random.choice([0, 1])

        # Missed ANC followups
        missedFollowups = random.randint(0, 3)

        # Area risk factor
        areaRiskIndex = round(random.uniform(0.1, 0.9), 2)

        # Risk Probability Logic

        risk_probability = 0.1

        if hemoglobin < 9.5:
            risk_probability += 0.35

        if hemoglobin < 10.5:
            risk_probability += 0.15

        if systolicBP > 135:
            risk_probability += 0.15

        if pastAnemia == 1:
            risk_probability += 0.20

        if missedFollowups >= 2:
            risk_probability += 0.15

        if fatigue == 1:
            risk_probability += 0.07

        if dizziness == 1:
            risk_probability += 0.05

        if areaRiskIndex > 0.65:
            risk_probability += 0.10

        # Controlled randomness
        risk_probability += random.uniform(-0.05, 0.05)

        developedHighRiskNextVisit = 1 if risk_probability > 0.5 else 0

        data.append([
            trimester,
            weight,
            hemoglobin,
            systolicBP,
            diastolicBP,
            dizziness,
            fatigue,
            swelling,
            lowAppetite,
            pastAnemia,
            missedFollowups,
            areaRiskIndex,
            developedHighRiskNextVisit
        ])

    columns = [
        "trimester",
        "weight",
        "hemoglobin",
        "systolicBP",
        "diastolicBP",
        "dizziness",
        "fatigue",
        "swelling",
        "lowAppetite",
        "pastAnemia",
        "missedFollowups",
        "areaRiskIndex",
        "developedHighRiskNextVisit"
    ]

    df = pd.DataFrame(data, columns=columns)

    return df
```

```python
df_pregnancy = generate_pregnancy_dataset(rows=500)
df_pregnancy.head(10)
```
