```python
import pandas as pd
import numpy as np
import random
```

```python
def generate_dataset(rows=500, seed=42):
    random.seed(seed)
    np.random.seed(seed)

    data = []

    for _ in range(rows):

        # Age (6–60 months)
        ageMonths = random.randint(6, 60)

        # Weight progression (approx realistic trend)
        weight = round(np.random.normal(0.24 * ageMonths + 4, 1.0), 2)

        # Height progression
        height = round(np.random.normal(60 + ageMonths * 0.5, 2.5), 2)

        # MUAC realistic range
        muac = round(np.random.normal(12.5, 1.0), 2)

        # Symptom Flags (Binary)
        weakness = random.choice([0, 1])
        lowAppetite = random.choice([0, 1])
        frequentIllness = random.choice([0, 1])
        diarrhea = random.choice([0, 1])

        # Past risk distribution
        pastRiskScore = random.choices([0,1,2], weights=[0.5,0.3,0.2])[0]

        # Visit gap
        visitGapDays = random.randint(7, 60)

        # Area risk index
        areaRiskIndex = round(random.uniform(0.1, 0.9), 2)

        # -------- Nutrition Need Logic --------
        if muac < 11.5:
            nutritionNeed = "energy_dense"

        elif weakness == 1:
            nutritionNeed = "iron_rich"

        elif frequentIllness == 1:
            nutritionNeed = "immunity_boost"

        elif lowAppetite == 1:
            nutritionNeed = "light_meals"

        else:
            nutritionNeed = "balanced"

        data.append([
            ageMonths,
            weight,
            height,
            muac,
            weakness,
            lowAppetite,
            frequentIllness,
            diarrhea,
            pastRiskScore,
            visitGapDays,
            areaRiskIndex,
            nutritionNeed
        ])

    columns = [
        "ageMonths",
        "weight",
        "height",
        "muac",
        "weakness",
        "lowAppetite",
        "frequentIllness",
        "diarrhea",
        "pastRiskScore",
        "visitGapDays",
        "areaRiskIndex",
        "nutritionNeed"
    ]

    df = pd.DataFrame(data, columns=columns)

    return df
```

```python
df= generate_dataset(rows=500)
df.head(10)
```
