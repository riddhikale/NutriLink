```python
import pandas as pd
import numpy as np
import random
```

```python
def generate_dataset(rows=500, seed=42):
    random.seed(seed)
    np.random.seed(seed)

    data = []

    for _ in range(rows):

        # Age (6–60 months)
        ageMonths = random.randint(6, 60)

        # Weight progression (approx realistic trend)
        weight = round(np.random.normal(0.24 * ageMonths + 4, 1.0), 2)

        # Height progression
        height = round(np.random.normal(60 + ageMonths * 0.5, 2.5), 2)

        # MUAC realistic range
        muac = round(np.random.normal(12.5, 1.0), 2)

        # Symptom Flags (Binary)
        weakness = random.choice([0, 1])
        lowAppetite = random.choice([0, 1])
        frequentIllness = random.choice([0, 1])
        diarrhea = random.choice([0, 1])

        # Past risk distribution
        pastRiskScore = random.choices([0,1,2], weights=[0.5,0.3,0.2])[0]

        # Visit gap
        visitGapDays = random.randint(7, 60)

        # Area risk index
        areaRiskIndex = round(random.uniform(0.1, 0.9), 2)

        # -----------------------------
        # Risk Probability Logic
        # -----------------------------

        risk_probability = 0.1  # base risk

        if muac < 11.5:
            risk_probability += 0.35

        if weight < (0.23 * ageMonths + 3.5):
            risk_probability += 0.25

        if visitGapDays > 35:
            risk_probability += 0.15

        if areaRiskIndex > 0.65:
            risk_probability += 0.10

        if pastRiskScore == 2:
            risk_probability += 0.20

        if weakness == 1:
            risk_probability += 0.08

        if frequentIllness == 1:
            risk_probability += 0.07

        if diarrhea == 1:
            risk_probability += 0.05

        # Controlled randomness
        risk_probability += random.uniform(-0.05, 0.05)

        # Final label
        becameHighNextVisit = 1 if risk_probability > 0.5 else 0

        data.append([
            ageMonths,
            weight,
            height,
            muac,
            weakness,
            lowAppetite,
            frequentIllness,
            diarrhea,
            pastRiskScore,
            visitGapDays,
            areaRiskIndex,
            becameHighNextVisit
        ])

    columns = [
        "ageMonths",
        "weight",
        "height",
        "muac",
        "weakness",
        "lowAppetite",
        "frequentIllness",
        "diarrhea",
        "pastRiskScore",
        "visitGapDays",
        "areaRiskIndex",
        "becameHighNextVisit"
    ]

    df = pd.DataFrame(data, columns=columns)

    return df
```

```python
df= generate_dataset(rows=500)
df.head(10)
```
